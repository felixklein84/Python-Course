**Create a function named domainGet that extracts the e-mail domain from the following string: **

    user@domain.com
    
**Here "user@domain.com" returns: domain.com**

```python
# define the function here
def domainGet(x):
    return x.split("@")[-1]
```

```python
domainGet('user@domain.com')
```

** Create a function with the name look_for_money which returns True if the word 'money' is contained in the input string. Capitalization should be considered, but not if a character is attached to the word (e.g. 'money,' is not considered). **

```python
# define function here
def look_for_money(x):
    return "money" in x.lower().split()
```

```python
look_for_money('I have so much money in my pockets!')
```

** Create a function count_money which counts the number of the word money in the input string. Capitalization should be taken into account, but not if a character is attached to the word (e.g. ‘money,’ is not taken into account). **

```python
def count_money(x):
    return sum([word in "money" for word in x.lower().split()])
```

```python
count_money('Money all I need is money Money, dude!')
```

**You are driving a little too fast and a police officer stops you. Write a function to return one of 3 possible results: “No ticket”, “Minor ticket” or “Major ticket”. If your speed is 100 or less, the result is “No ticket”. If the speed is between 101 and 120 inclusive, the result is “Small ticket”. If the speed is 121 or more, the result is “Big Ticket”. Unless it is your birthday (coded as a boolean value in the parameters of the function) -- on your birthday, your speed can be 10 higher in all cases. **

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 10
    if speed <= 100:
        return "No Ticket"
    elif speed <= 120:
        return "Small Ticket"
    else:
        return "Big Ticket"
```

```python
caught_speeding(101,True)
```

```python
caught_speeding(101,False)
```
