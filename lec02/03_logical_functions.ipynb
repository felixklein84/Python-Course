## if,elif, else

```python
if 1 < 2:
    print('Yep!') # Spacing important (Python does not use {}); is automatically inserted by Jupyter and most IDEs
```

```python
if 1 < 2:
    print('yep!')
```

```python
if 1 < 2:
    print('first')
else:
    print('last')
```

```python
if 1 > 2:
    print('first')
else:
    print('last')
```

```python
# you can use as many elif statements as you like, the block of the first valid statement is always executed
if 1 == 2:
    print('first')
elif 3 == 3:
    print('middle')
else:
    print('Last')
```

## for loops

```python
seq = [1,2,3,4,5]
```

```python
# allows you to iterate through a list/sequence
for item in seq:
    print(item)
```

```python
for item in seq:
    print('Yep')
```

```python
for num in seq:
    print(num+num)
```

## while loops

```python
i = 1
while i < 5:
    print('i is: {}'.format(i))
    i = i+1 # This line is very important, otherwise we will create an infinite loop
```

## list comprehension

```python
x = [1,2,3,4]
```

```python
out = []
for item in x:
    out.append(item**2)
print(out)
```

```python
# in Python there is the possibility to shorten the code of the previous cell
# very useful concept
[item**2 for item in x]
```

## lambda expression

```python
def times2(var):
    return var*2
```

```python
times2(2)
```

```python
lambda var: var*2
```

## map and filter

```python
seq = [1,2,3,4,5]
```

```python
map(times2,seq)
```

```python
list(map(times2,seq))
```

```python
list(map(lambda var: var*2,seq))
```

```python
filter(lambda item: item%2 == 0,seq)
```

```python
list(filter(lambda item: item%2 == 0,seq))
```
