## Simple Initializations

```python
noten = [1, 2, 3, 4, 5, 6]
# Create a NumPy vector from the list
noten_as_vector = np.array(noten)
```

```python
# Now create a matrix with 1,2,...,6 on the diagonal and 0 everywhere else by multiplying the identity matrix by the vector of grades.
einheitsmatrix * noten
```

## Random Numbers and Simple Operations

```python
# Initialisiere eine Zufallszahlengenerator mit np.random.default_rng()
zufallszahlengenerator =  np.random.default_rng()
```

```python
# Create a 6x6 matrix of normally distributed values with mean 2.5 and variance 1
normalmatrix = zufallszahlengenerator.normal(2.5, 2, (6,6))
print(normalmatrix)
```

```python
# Now multiply the matrix by the exponential function. The resulting random variables will then follow a log-normal distribution.
np.exp(normalmatrix)
```

```python
# Display row 6 and columns 2-3 of normalmatrix
print(normalmatrix[5, 1:3])
```

## And finally, here's a slightly more challenging task

```python
# Display all entries of normalmatrix that are above the empirical mean of all observations
normalmatrix_larger = normalmatrix[normalmatrix>np.mean(normalmatrix)]
print(normalmatrix_larger)
```

```python
# Now calculate the mean and median of the previously selected elements
print(np.mean(normalmatrix_larger))
print(np.median(normalmatrix_larger))
```
